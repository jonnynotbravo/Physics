# 🧠 Physics Phase 1: Foundations of Physics (Conceptual, No Calculations)


## 🔹 1. What is Physics?

**Physics** is the study of how the universe works, from the smallest particles to the largest structures in the cosmos. It’s the foundation of all natural sciences.

### 🔬 Branches of Physics:
- **Classical Mechanics** – motion, forces
- **Thermodynamics** – heat and energy
- **Electromagnetism** – light, electricity, magnetism
- **Relativity** – time, space, gravity (Einstein)
- **Quantum Mechanics** – atoms, subatomic particles
- **Nuclear Physics** – atomic nuclei and radiation
- **Particle Physics** – fundamental particles
- **Condensed Matter** – solids, liquids, superconductors
- **Astrophysics/Cosmology** – stars, galaxies, universe

### 💡 Why It Matters:
Physics explains everything from how apples fall to why stars shine — it’s the **source code of reality**.



## 🔹 2. Scientific Method

A systematic process for uncovering truth:

1. **Observation** – something interesting happens  
2. **Question** – ask why or how  
3. **Hypothesis** – educated guess  
4. **Prediction** – what should happen  
5. **Test** – experiment or collect data  
6. **Analysis** – compare results  
7. **Conclusion** – revise or confirm

### ⚠️ Key Point:
Even successful theories are never “proven” — just **strongly supported**.



## 🔹 3. The Four Fundamental Forces

All forces in the universe reduce to these four:

| Force             | What it Affects         | Strength     | Range       | Carrier Particle       |
|------------------|--------------------------|--------------|-------------|------------------------|
| **Gravity**       | Mass                     | Weakest      | Infinite    | Graviton (theoretical) |
| **Electromagnetism** | Electric charge       | Medium-Strong| Infinite    | Photon                 |
| **Strong Nuclear** | Quarks (holds nuclei)   | Strongest    | Tiny        | Gluon                  |
| **Weak Nuclear**   | Subatomic particles     | Weak         | Very tiny   | W and Z bosons         |

### ✅ Everyday Forces (Friction, Tension, etc.) are **not fundamental**  
They emerge from particle-level interactions of the fundamental forces.



## 🔹 4. What is a Force?

A **force** is a **push or pull** that changes an object’s **motion, speed, or shape**.

### Formula (for context):
## \[ F = ma ] 

Where:
- \(F\) = force (newtons)
- \(m\) = mass
- \(a\) = acceleration

### 🧲 Forces are Interactions:
They always happen between **at least two objects**.  
One exerts the force, one experiences it — but both are affected equally (Newton’s 3rd Law).



## 🔹 5. SI Units (International System of Units)

Used to make measurements **universal and meaningful**.  
Each number must be paired with a **unit** to give context.

### 7 Fundamental SI Units:

| Quantity               | Unit    | Symbol |
|------------------------|---------|--------|
| Length                 | meter   | m      |
| Mass                   | kilogram| kg     |
| Time                   | second  | s      |
| Electric current       | ampere  | A      |
| Temperature            | kelvin  | K      |
| Amount of substance    | mole    | mol    |
| Luminous intensity     | candela | cd     |

### 🌟 Derived Units:
- Speed → m/s  
- Force → newton (N) = kg·m/s²  
- Pressure → pascal (Pa) = N/m²  
- Energy → joule (J) = kg·m²/s²



## 🔹 6. Orders of Magnitude

### Definition:
**Order of magnitude** = how many powers of 10 something has

### Example:
- Human = \(10^0\) m
- Cell = \(10^{-6}\) m
- Earth = \(10^7\) m
- Sun = \(10^9\) m

Each jump in exponent is **10x bigger or smaller**.

### Negative powers (e.g., \(10^{-3}\)) = smaller than 1  
Positive powers = larger than 1



## 🔹 7. The Physics Mindset

### Key Principles:

- **Models**: Simplified versions of reality (not perfect, but useful)
- **Conservation Laws**:  
 Energy, momentum, and charge are always conserved
- **Symmetries**:  
 If physics stays the same over time or space → something is conserved (like energy or momentum)
- **Skepticism + Curiosity**:  
 Don’t assume — always test, observe, and revise
